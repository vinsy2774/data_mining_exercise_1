## Data Preparation

Missing Values Repair

```python
import pandas as pd

original_data = pd.read_csv('../data/titanic.csv')

copy_data = original_data.copy()

copy_data.head(5)
```

```python
missing_value_columns = ['Age', 'Cabin', 'Embarked']

for col in missing_value_columns:
    print(copy_data[col].isna().value_counts())
```

```python
# Removing Cabin Column from the dataset as it has too many missing values

copy_data.drop('Cabin', axis=1, inplace=True)
```

```python
copy_data.head(5)
```

```python
# Assigning the mean value of all the ages for the missing age values

not_missing_values = copy_data[copy_data['Age'].isna() == False]
Average_age = int(not_missing_values['Age'].astype('int').mean())

copy_data['Age'] = copy_data['Age'].fillna(Average_age)

```

```python
copy_data.head(5)
```

```python
# Dropping remaining missing values rows

copy_data.dropna(inplace=True)
```

Renaming the columns

```python
# Renaming the 'SibSp' and 'Parch' columns

copy_data.rename(columns= {
    'SibSp' : 'No_of_Siblings_and_Spouses_Aboard',
    'Parch' : 'No_of_Parents_and_Children_Aboard'
},inplace=True)
```

```python
copy_data.columns
```

Renaming column values

```python
copy_data['Embarked'].value_counts()
```

```python
# Renaming the Embarked column values to meaningful ones

copy_data.replace({'Embarked': {
    29 : 'Q'
}}, inplace=True)
copy_data['Embarked'].value_counts()
```

```python
copy_data.replace({'Embarked' : {
    'S' : 'Southampton',
    'Q' : 'Queenstown',
    'C' : 'Cherbourg'
}}, inplace=True)
copy_data['Embarked'].value_counts()
```

```python
copy_data.head(5)
```

```python
# Adding a new column : Underage - true for passengers under 18 and False for passengers over 18

copy_data['Underage'] = copy_data['Age'] < 18
```
