## Visualising the data

```python
import pandas as pd 

copy_data = pd.read_pickle('../notebooks/titanic.pkl')
```

```python
copy_data.head(5)
```

Survival of Passengers Under 25

```python
import matplotlib.pyplot as plt 


labels = ['dead', 'survived']

Under_25 = copy_data[copy_data['Age'] < 25]

Survived_or_dead = Under_25['Survived'].value_counts()

sizes = [Survived_or_dead.get(0,0), Survived_or_dead.get(1,0)]

fig, ax = plt.subplots()
ax.pie(sizes, labels=labels, autopct='%1.1f%%', colors=['Red', 'Green'])
plt.title('Survival of Passengers Under 25')
plt.Figure(figsize=(7,7))
plt.show()
```

```python
Ages = copy_data['Age'].values

fig, ax = plt.subplots(figsize=(7,7))
ax.hist(Ages, bins=5, linewidth=0.5, edgecolor='white')
ax.set_xlabel("Age")
ax.set_ylabel("Frequency")
plt.title('Histogram of Ages')
plt.show()
```

```python
Age = copy_data['Age']
Survival = copy_data['Survived']

fig, ax = plt.subplots(figsize=(7,7))
ax.scatter(Age, Survival, alpha=0.5, c='blue')
plt.title('Relationship between Age and Survival')
plt.xlabel('Age')
plt.ylabel('Survival (0 - No, 1 - Yes)')
ax.set_yticks([0,1])
ax.set_ylim(-0.2, 1.2)
ax.grid(True)
plt.show()
```

No, there is no relation betweeen Age and Survival
