Oldest Surviving Male & Female Passenger

```python
copy_data['Survived'] = original_data['Survived']

survived_passengers = copy_data[copy_data['Survived'] == 1]

male_survived_passengers = survived_passengers[survived_passengers['Sex'] == 'male']
female_survived_passengers = survived_passengers[survived_passengers['Sex'] == 'female']

oldest_male_survived_passenger = male_survived_passengers.loc[male_survived_passengers['Age'].idxmax(), ['Name','Age']]
oldest_female_survived_passenger = female_survived_passengers.loc[female_survived_passengers['Age'].idxmax(), ['Name','Age']]
```

```python
print(f'Oldest Male Passenger ------ {oldest_male_survived_passenger['Name']} : {oldest_male_survived_passenger['Age']}')
print(f'Oldest Female Passenger ------ {oldest_female_survived_passenger['Name']} : {oldest_female_survived_passenger['Age']}')
```

Grouping the Passengers by Sex

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.expand_frame_repr', False)

grouped_data = copy_data.groupby('Sex')

for sex, group in grouped_data:
    print(f'Rows for Sex: {sex}')
    print(group.head(5))
```

Grouping the Passengers by Pclass

```python
grouped_data = copy_data.groupby('Pclass')

for Pclass, group in grouped_data:
    print(f'Rows for Passenger Class: {Pclass}')
    print(group.head(5))
```
